# Extract and Transform 2025 NYC Yellow Taxi Data

## Extract

```python
import polars as pl
```

```python
df = pl.scan_parquet("*.parquet")
```

## Transform

```python
df = df.with_columns(replaced=pl.col("payment_type").replace_strict([0, 1, 2, 3, 4, 5, 6], 
["FlexFareTrip", "CreditCard", "Cash", "NoCharge", "Dispute","Unknown","VoidedTrip"])
).drop("payment_type"
).rename({"replaced": "payment_type"})

df = df.with_columns(replaced=pl.col("RatecodeID").replace_strict([1, 2, 3, 4, 5, 6, 99], 
["StandardRate", "JFK", "Newark", "NassauOrWestchester", "NegotiatedFare", "GroupRide", "NullUnknown"])
).drop("RatecodeID"
).rename({"replaced": "RateCode"})

df = df.drop_nulls()
```

```python
df.sink_csv("2025_yellow_cab_data.csv")
```
